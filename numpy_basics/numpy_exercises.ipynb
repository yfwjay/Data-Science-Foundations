```python
import numpy as np
```

```python
# Create a numpy array of size 10 filled with zeros

a = np.zeros(10)
a
```

```python
# Create a numpy array with values ranging from 10 to 49

b = np.arange(10 , 59)
b
```

```python
# Create a numpy matrix of 2 * 2 integers filled with ones

a_ones = np.ones( [2 , 2] , dtype= 'int64')
a_ones
```

```python
# Create a numpy matrix of 3 * 2 float numbers , filled with ones

# b_ones = np.ones([3 , 2] , dtype = np.float64) 
b_ones = np.ones([3 , 2] , dtype= 'float')
b_ones
```

```python
# Given the X numpy array , create a numpy array with the same shape and type as X , filled with ones

x_ones = np.arange(4)

np.ones_like(x_ones)

print(f"X Array = {x_ones} \nOne Array = {np.ones_like(x_ones)}")
```

```python
# Given the X numpy matrix, create a new numpy matrix with the same shape and type as X, filled with zeros.

x_matrix = np.array([[2 , 3 , 4] , [5 , 6 ,7]])
np.zeros_like(x_matrix)

print(f"X Array = {x_matrix} \nOne Array = {np.zeros_like(x_matrix)}")
```

```python
# Create a numpy matrix of 4*4 integers, filled with fives.

five_array = np.ones([4 , 4]  , dtype= 'int64') *5
five_array
```

```python
# Given the X numpy matrix, create a new numpy matrix with the same shape and type as X, filled with sevens.

seven_array = np.array([[5 , 5] , [ 3 , 4]])

np.ones_like(seven_array) * 7

print(f"Array = {seven_array} \nSeven_Array = {np.ones_like(seven_array) * 7}")
```

```python
# Create a 3*3 identity numpy matrix with ones on the diagonal and zeros elsewhere.

identity_matrix = np.identity(3)
identity_matrix

similar_identity = np.eye(3)
similar_identity

print(f"Identity Matrix = {identity_matrix} \nSimlar Matrix = {similar_identity}")
```

```python
# Given the X python list convert it to an Y numpy array

x = [1 , 2 , 3]
x

x_array = np.array(x)
x_array

print(f"List = {x} and Datatype = {type(x)}" )
print(f"Array = {x_array} and Datatype = {type(x_array)}")
```

```python
# Given the X numpy array, make a copy and store it on Y.

new_x_array = np.array(4)

y = np.copy(new_x_array)

print(f"Array_x = {new_x_array} \nY = {y}")
```
