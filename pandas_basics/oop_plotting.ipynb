```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# OOP Interface Plotting

# This is the modern way of creating plots in matplot-lib

# fig = is like the entire window or canvas where you are going to draw. It deals with stiff like background color , saving image to a file

# axes = This is the individual plot in the graph. Here we set things like the title , draw data label the x and y axis , legend etc

fig , axes = plt.subplots(figsize = (12 , 6))


# Normally we used to say plt.plot() the global way

# using fig , axes gives you more control
```

```python
x = np.arange(-10 ,11)
```

```python
# We can draw the 1st axis

axes.plot( x , (x ** 2) , color = 'red' , linewidth = 3 , marker = 'o' , markersize = 8 , label = 'x^2')
axes.plot( x , -1 * (x ** 2) , color = 'blue' , linewidth = 3 , label = '-x^2')

# axes.plot( x , -1 * (x ** 2) , 'b--' , label = '-x^2')

axes.set_xlabel('X') # x_label
axes.set_ylabel('X Squared') # y-label
axes.set_title("My Nice Plot") # title

axes.legend() #we call the legend
fig
```

```python
# We can draw another figure like a line-plot

fig , axes = plt.subplots(figsize = (12 ,6))

# Create different line plots.(dashed , solid , dashdot , dotted)

axes.plot(x , x+0 , linestyle = 'solid')
axes.plot(x  , x+1 , linestyle = 'dashed')
axes.plot(x , x+2 , linestyle = 'dotted')
axes.plot(x , x+3 , linestyle = 'dashdot')
axes.set_title("Different Line Plots")
fig
```

```python
# # # SCATTER PLOT AND BUBBLE CHART

# # # 1. SCATTER CHART
# # # -> This is used to show the relationship between 2 numerical values. We show:
# #         # See relationships or correlations (positive, negative, none)

# #         Spot patterns or clusters

# #         Detect outliers

# #         Compare two continuous variables

# # EXAMPLE: Relationship between male and female , How height relates to weight.


# # 2. BUBBLE CHART

# # -> This is a scatter plot only that it has a 3rd variable. 
# # -> The size of each point(Bubble) represents another value.
# #  -> We use it when we want to see the relationship between 3 variables all at once.

# # FOR EXAMPLE: Student marks vs attendance vs study hours. (HOW DO THE THREE RELATE)


# # Different attributes in scatter(Bubble chart)

#         1. s = size of the bubble
#         2. c = color of the points or we can use values to represent the color
#         3. cmap = color mapping(Its like a ph scale for the c) common color map
#                         'viridis'
#                         'plasma'
#                         'coolwarm'
#         4. alpha = transperency of the bubbles (Range 0(invisible) and 1 (solid) Very useful when the bubbles overlap)
#         5. marker = This is the shape of the points.
#                         'o' = circle
#                         'x' = cross 
#                         '^' = triangle
#                         's' = square
#         6. edgecolors = This is the border color . Makes bubbles clearer especially with transperency
#                         'black' 


# example: A data set with columns student , study hours , Score , Attendance.
# we can make x to be study hours , y = scores and c = attendance
```

```python
# SCATTER PLOT

# We generate our values using numpy which were gonna plot

N = 50
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = np.pi * (20 * np.random.rand(N))**2  # 0 to 15 point radii


# We now plot

fig , axes = plt.subplots(figsize = (14 , 6))

scatter = axes.scatter(x , y , s = area , c = colors , alpha=0.5 , cmap= 'Spectral')

# From our figure we dont have the color bar so we must specify it

fig.colorbar(scatter , ax = axes , label = 'color Values')
```

```python
# We try and impliment the same but now we have one row , 2 subplots

fig , ((ax1 , ax2)) = plt.subplots(nrows=1, ncols=2,figsize = (14 , 6))

scatter1 = ax1.scatter(x , y , s = area , c = colors , cmap= 'Pastel1' , alpha= 0.6)
scatter2 = ax2.scatter(x , y , s = area , c = colors , cmap= 'Spectral' , alpha= 0.6)

fig.colorbar(scatter1, ax=ax1, label='Pastel1 values')
fig.colorbar(scatter2, ax=ax2, label='Spectral values')
```

```python
# # HISTOGRAM

# A histogram is used to show the distribution of a single numerical variable. 
# Like if it is a number like 1 we look like how many time has it occured
# The x-axis = Value ranges(bins)
# The y-axis = frequency (count)

# Typical questions histograms answer
# How are students’ marks spread?
# Do most values cluster around the mean?
# Is the data left-skewed or right-skewed?
# Are there multiple peaks?


# HISTOGRAM COMMON ATTRIBUTES

# 1.bins = number of intervals.
#                 Fewer bins → smoother, less detail
#                 More bins → more detail, more noise
# The smaller the dataset the fewer the bins and viceversa

# 2.color = bar color. eg: skyblue

# 3.edgecolor = border of the bars for easier identification. eg: black ,blue etc
        
# 4.alpha = transperency of the bars.

# 5.density = probability instead of  frequency density = True means the y-axis becomes probabiliy density instead of frequency

# 6.histtype = style of the histogram. EG: bar , step , stepfilled

# 7.orientation = horizontal histogram
```

```python
# # HISTOGRAM ANSWERS 5 QUESTIONS

# 1.Center
    # Look where the tallest bars are clustered. 
    # In our case the tallest bars re clustered around bar 0. HENCE: The data is centered around 0 (mean ≈ 0)

# 2.Spread
    # Look how far the bars extend towards the left and the right.
    # In our case values fall between -3 and 3. HENCE: Most values are within a limited range, indicating moderate variability.

# 3.Shape
    # In our case we should see a bell-shaped curve. THat means:
    # Normal distribution or One main Peak. HENCE: “The histogram shows a unimodal, approximately normal distribution.
    # Uniform (flat)

    # Bimodal (two peaks)

    # Skewed (tail on one side)

# 4.Skewness
    # Look at the tails:

    # If right tail is longer → right skewed

    # If left tail is longer → left skewed

    # If both sides are similar → symmetric

    # IN OUR CASE: Both the left and right tails are roughly equal hence symetric

# 5.Outliers

    # These are extreme or rare values
    # Look for isolated bars very far away from the center OR
    # Very low frequency bins at the extremes
    # IN OUR CASE: A few values near ±3 or more.
```

```python
data = np.random.randn(1000)

fig , axes = plt.subplots(figsize = (14 ,5))

axes.hist(data , bins= 100 , alpha = 0.8 , histtype='bar' , color='steelblue' , edgecolor = 'black' ,)
```
