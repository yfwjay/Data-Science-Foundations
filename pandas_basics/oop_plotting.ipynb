```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# OOP Interface Plotting

# This is the modern way of creating plots in matplot-lib

# fig = is like the entire window or canvas where you are going to draw. It deals with stiff like background color , saving image to a file

# axes = This is the individual plot in the graph. Here we set things like the title , draw data label the x and y axis , legend etc

fig , axes = plt.subplots(figsize = (12 , 6))


# Normally we used to say plt.plot() the global way

# using fig , axes gives you more control
```

```python
x = np.arange(-10 ,11)
```

```python
# We can draw the 1st axis

axes.plot( x , (x ** 2) , color = 'red' , linewidth = 3 , marker = 'o' , markersize = 8 , label = 'x^2')
axes.plot( x , -1 * (x ** 2) , color = 'blue' , linewidth = 3 , label = '-x^2')

# axes.plot( x , -1 * (x ** 2) , 'b--' , label = '-x^2')

axes.set_xlabel('X') # x_label
axes.set_ylabel('X Squared') # y-label
axes.set_title("My Nice Plot") # title

axes.legend() #we call the legend
fig
```

```python
# We can draw another figure like a line-plot

fig , axes = plt.subplots(figsize = (12 ,6))

# Create different line plots.(dashed , solid , dashdot , dotted)

axes.plot(x , x+0 , linestyle = 'solid')
axes.plot(x  , x+1 , linestyle = 'dashed')
axes.plot(x , x+2 , linestyle = 'dotted')
axes.plot(x , x+3 , linestyle = 'dashdot')
axes.set_title("Different Line Plots")
fig
```
