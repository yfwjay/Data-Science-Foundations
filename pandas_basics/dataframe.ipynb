```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Create an empty data frame

pd.DataFrame()
```

```python
# Create a marvel_df pandas DataFrame with the given marvel data

marvel_data = [
    ['Spider-Man', 'male', 1962],
    ['Captain America', 'male', 1941],
    ['Wolverine', 'male', 1974],
    ['Iron Man', 'male', 1963],
    ['Thor', 'male', 1963],
    ['Thing', 'male', 1961],
    ['Mister Fantastic', 'male', 1961],
    ['Hulk', 'male', 1962],
    ['Beast', 'male', 1963],
    ['Invisible Woman', 'female', 1961],
    ['Storm', 'female', 1975],
    ['Namor', 'male', 1939],
    ['Hawkeye', 'male', 1964],
    ['Daredevil', 'male', 1964],
    ['Doctor Strange', 'male', 1963],
    ['Hank Pym', 'male', 1962],
    ['Scarlet Witch', 'female', 1964],
    ['Wasp', 'female', 1963],
    ['Black Widow', 'female', 1964],
    ['Vision', 'male', 1968]
]

marvel_df = pd.DataFrame(marvel_data)
marvel_df
```

```python
# Add column names to the marvel_df

col_names = ['name', 'sex', 'first_appearance']

marvel_df.columns = col_names

marvel_df
```

```python
# Add index names to the marvel_df (use the character name as index)

marvel_df.index = marvel_df['name']
marvel_df
```

```python
# # Drop the name column as it's now the index

marvel_df = marvel_df.drop(columns = ['name'])
```

```python
# Drop 'Namor' and 'Hank Pym' rows

marvel_df = marvel_df.drop(['Namor' , 'Hank Pym'])
```

```python

# DataFrame boolean arrays (also called masks)

```

```python
# Given the marvel_df pandas DataFrame, make a mask showing the female characters

selected = marvel_df['sex'] == 'female'
marvel_df[selected]
```

```python
# Given the marvel_df pandas DataFrame, get the male characters

male_characters = marvel_df['sex'] == 'male'

marvel_df[male_characters]
```

```python
# Given the marvel_df pandas DataFrame, get the characters with first_appearance after 1970

after_1970 = marvel_df['first_appearance'] > 1970

marvel_df[after_1970]
```

```python
# Given the marvel_df pandas DataFrame, get the female characters with first_appearance after 1970

female_after_1970 = (marvel_df['sex'] == 'female') & (marvel_df['first_appearance'] > 1970)

marvel_df[female_after_1970]
```
