```python
import pandas as pd
import sqlite3
import numpy as np
```

```python
# Crypto currency database
# The database we will use contains two tables cryptocoins_cryptocurrency and cryptocoins_exchange.

# Using the given sqlite3 connection and the read_sql method write a single query that:

# Join both tables and return the list of cryptocurrencies with its exchanges. You should use the column with the exchange ID on each table to perform the join.

# As both tables have a name column you should rename cryptocoins_cryptocurrency.name as coin_name and cryptocoins_exchange.name as exchange. Also select symbol, price_usd and percent_change_7d columns.

# Store the information in a crypto_df DataFrame.

# Once you have the crypto_df DataFrame, create a new weekly_change_df with the crypto_df data sorted by percent_change_7d from highest to lowest.
```

```python
# This note book will be using the cryptos-sql file. W need to read it and execute it

# 1. Craete an empty database to our ram

conn = sqlite3.connect(':memory:')

# 2. We open our sql file

with open(r'C:\Users\HP\Desktop\Data-Science-Foundations\Databases\cryptos.sql' , 'r') as file:
    sqlite3_script = file.read()

# We execute our sql script

conn.executescript(sqlite3_script)

# We create a 2 dfs. The sql script has many tables but we will be using 2 tables.(cryptocoins_cryptocurrency and cryptocoins_exchange)

df_crypto_currency= pd.read_sql('SELECT * FROM cryptocoins_cryptocurrency' , conn , index_col=['id'])

# 2ND DF

df_crypto_exchange = pd.read_sql("SELECT * FROM cryptocoins_exchange" , conn)
```

```python
df_crypto_currency
```

```python
df_crypto_currency = df_crypto_currency.rename(columns={'name' : 'coin_name'})
```

```python
df_crypto_exchange = df_crypto_exchange.rename(columns={'name' : 'exchange'})
```

```python
df_crypto_exchange
```

```python
# Merge both dfs

combined_df = pd.merge(df_crypto_currency , df_crypto_exchange , left_on= 'exchange_id' , right_on= 'id' , how= 'inner')

combined_df.drop(columns = ['id'])
```

```python
crypto_df = combined_df.loc[:: , ('symbol' , 'price_usd' , 'percent_change_7d')]
crypto_df
```

```python
# Once you have the crypto_df DataFrame, create a new weekly_change_df with the crypto_df data sorted by percent_change_7d from highest to lowest.

weekly_change_df = crypto_df.sort_values(by='percent_change_7d', ascending=False)
weekly_change_df
```
