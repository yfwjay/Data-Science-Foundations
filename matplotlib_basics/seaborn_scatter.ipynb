```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read one csv file

df = pd.read_csv(r'C:\Users\HP\Desktop\Data-Science-Foundations\datasets\social_media_engagement_dataset.csv')

df
```

```python
# W ecan use a countplot and visualize the platform column
fig , axes = plt.subplots(figsize = (12 , 6))

hue_colors = {'Text' : 'Green' , 'Video' : 'Blue' , 'Image' : 'Yellow'}

sns.countplot(x = 'platform' , hue = 'post_type' ,palette= hue_colors , data = df)
```

```python
# The dataframe has a lot of rows so we may use 100 rows, But in random

# We use a seed so that the same rows are selected every time

sample_df = df.sample(n = 100 , random_state= 42)
sample_df
```

```python
# RELATIONAL PLOT

# Used to show the realtionship between 2 numerical variables(x and y) using either a scatter plot or linear plot.

```

```python
# A relational plot of kind scatter

# We will be checking the relationship between a post length and views and plot each scatter plot for each of column post_type

# Each post type has their own scatter plot
fig , axes = plt.subplots(figsize = (12 , 6))

sns.relplot(x = 'post_length' , y = 'views' , kind= 'scatter' , ax = axes , data = sample_df, col= 'post_type')

# Also we can specify the col_order for post_type. It has text , video and image. We may specify the order by passing a list.

# col_order = ['video' , 'image' , 'text']
```

```python
# ADDED 2 ATTRIBUTES : SIZE AND HUE

# SIZE = it eill create a legend with dots of different sizes to help you interprate the data. But of the same color

# hue = This allows us to have different colors in the dots of the scatter plot

fig , axes = plt.subplots(figsize = (12 , 6))

sns.relplot(x = 'post_length' , y = 'views' , kind= 'scatter' , ax = axes , data = sample_df , size= 'post_type' , hue = 'post_type')
```

```python
# ATTRIBUTE: STYLE

# style: Instead of having the same style(circle) we change this by using the style option

fig , axes = plt.subplots(figsize = (12 , 6))

sns.relplot(x = 'post_length' , y = 'views' , kind= 'scatter' , ax = axes , data = sample_df , style= 'post_type' , hue= 'post_type')
```
