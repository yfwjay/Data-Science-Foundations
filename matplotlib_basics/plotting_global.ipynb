```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Matplotlib pyplot has 2 common APIS:

# Global Api = Matplot lib interface. We follow the universal way of plooting. Its common across everywhere
# OOP Interface = Its the one you specify yourself how you want to plot
```

```python
# Generate random numbers from -10 to 10
x = np.arange(-10 , 11)
x
```

```python
# PLOTTING USING THE GLOBAL API(THE COMMON WAY)
```

```python
# We then try to plot

# 1. We create the figure 

plt.Figure(figsize=(12 , 6))

# We give it a title

plt.title("Nice Plot")

# Using the above we can now plot inside our figure
# 1st parameter is what to have on the x axis and second parameter is what to have on the y-axis
plt.plot( x , x ** 2)

# We can plot anothere figure(same curve but now upside down)
plt.plot(x , -1 * (x ** 2))
```

```python

# As you can see from above we only have one figure. What if we wanted to have multiple figures = what we call sub-plots


```

```python
# We redo the same we create our figure and assign the title as the 1st step

plt.Figure(figsize = (20 , 14) )
plt.title("My Nice Plot")


# Now to have multiple figures we use a sub plot and specify row , column and fig parameters respectively

plt.subplot(1, 2, 1)  # rows, columns, panel selected

# 1. TOtal number of rows
# 2. Total number of columns
# 3. The panel selected(selects panel on the left side to be the active one)

plt.plot(x, x ** 2) # we draw the 1st figure what we had
plt.plot([0 , 0 , 0] , [-10 , 0 , 100]) # A clever way to draw a vertical line


# We now label our fu=igure the xaxis , y-axis , legend

plt.legend(['x^2' , 'vertical_line']) #You techically give akey. The order should be how you plotted what came 1st.
plt.xlabel('x')
plt.ylabel('X-Squared')


# WE NOW DO THE SAME FOR THE SECOND PLOT

plt.subplot(1 , 2 , 2)
plt.plot(x , -1 *(x ** 2))
plt.plot([-10, 0, 10], [-50, -50, -50])
plt.legend('-x^ 2' , 'Horizontal-line')
plt.xlabel('x')
plt.ylabel('x-squared')
```
