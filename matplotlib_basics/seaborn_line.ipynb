```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read one csv file

df = pd.read_csv(r'C:\Users\HP\Desktop\Data-Science-Foundations\datasets\social_media_engagement_dataset.csv')

df
```

```python
# The dataframe has a lot of rows so we may use 100 rows, But in random

# We use a seed so that the same rows are selected every time

sample_df = df.sample(n = 100 , random_state= 42)
sample_df
```

```python
# LINE PLOT

# We show the relationship between 2 numerical values by connecting data points with lines.
```

```python
# Has the same attributes as that of the scatter plot:
# x_ axis
# y_axis
# hue
# markers
# col
# row
# col_order
# row_order
# error bar
# style
# dashes
# alpha

fig , axes = plt.subplots(figsize = (12 , 6))

g = sns.relplot(x = 'views' , y = 'comments' , kind= 'line' , style= 'platform' ,markers = True , dashes = False , hue = 'platform', ax = axes , data = sample_df)

g.fig.suptitle("A line Plot showing how comments and Views vary across different platforms" , fontsize = 12)

# Adjust the title so that it doesnt cross t the plots

g.fig.subplots_adjust(top= 0.9)
```
