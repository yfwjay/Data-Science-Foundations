```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
# We will read one of our data sets. Most likely the social media one. It has categorical vraibles.

df = pd.read_csv(r"C:\Users\HP\Desktop\Data-Science-Foundations\datasets\social_media_engagement_dataset.csv")
df
```

```python
# since it has alot of rows we will use 100. So we select randomly 100 rows and use seed so that we make sure each time we run the code we always get the same rows

sample_df = df.sample(n = 100 , random_state= 42)
sample_df
```

```python
# 2. Bar chart

# We plot one of our columns

fig , axes = plt.subplots(figsize = (12 , 6))

c = sns.catplot(x = 'post_type' , y = 'views' , data = sample_df , kind = 'bar' , hue = 'platform')

c.fig.suptitle("A Bar Chart showing the Post Type Per Views on each Platform Present" , fontsize = 12)

# Position our title to a nice position to prevent it from overlapping from the plot

c.fig.subplots_adjust(top = 0.9)
```
